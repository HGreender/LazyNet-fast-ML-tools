```python
from fast_ml_tools import LazyNet


BASE_DIR = r'path_to_directory/'

model = LazyNet(
    model_name = 'efficientnetb4_unet',
    model_path = None,
    loss_name = 'dice_focal',
    train_img_dirs = BASE_DIR + 'orig/',
    train_mask_dirs = BASE_DIR + 'prepocessed_masks/',
    val_img_dirs = None,
    val_mask_dirs = None,
    train_ratio = 0.85,
    train_val_seed = 24,
    early_stopping_threshold = 15,
    epochs = 10,
    batch_size = 10,
    lr = 1e-4,
    lr_patience = 5,
    lr_factor = 0.25,
    verbose = True,
    device_id = 0,
    num_workers = 0,
    classes = ['target'],
    metric_names = None
)
```

```python
model.draw_logs()
```

```python
model.predict_single(
    image_path = './data/image.png',
    threshold = 0.5,
    visualize = True,
    save_concat_path = './data/mask_concat.png',
    save_mask_path = './data/mask.png'
)
```

```python
model.predict_folder(
    input_dir = './data/images/',
    output_dir = './data/predictions/',
    threshold = 0.5,
    save_masks = True,
    save_concat = True,
    visualize_console = False,
    mask_suffix = '_pred',
    viz_suffix = '_viz'
)
```
