```python
from fast_ml_tools import LazyNet


BASE_DIR = r'path_to_directory/'

model = LazyNet(
    model_name = 'efficientnetb4_unet',
    model_path = None,
    loss_name = 'dice_focal',
    train_img_dirs = BASE_DIR + 'orig/',
    train_mask_dirs = BASE_DIR + 'prepocessed_masks/',
    val_img_dirs = None,
    val_mask_dirs = None,
    train_ratio = 0.85,
    train_val_seed = 24,
    early_stopping_threshold = 15,
    epochs = 10,
    batch_size = 10,
    lr = 1e-4,
    lr_patience = 5,
    lr_factor = 0.25,
    verbose = True,
    device_id = 0,
    num_workers = 0,
    classes = ['target'],
    metric_names = None
)
```

```python
# Начните обучение и принудительно остановите до окончания обучения
model.fit(
    save_model_path='./test/check.pth', 
    save_logs_path='./test/check.json', 
    save_checkpoint_path='./test/checkpoint.pth'
)
```

```python
# При запуске этой ячейки обучение продолжится с того же места, где остановилось
model.fit(
    save_model_path='./test/check.pth', 
    save_logs_path='./test/check.json', 
    save_checkpoint_path='./test/checkpoint.pth',
    load_checkpoint_path='./test/checkpoint.pth'
)
```
